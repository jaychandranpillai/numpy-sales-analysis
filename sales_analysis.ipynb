# NumPy Sales Analysis
5 products × 6 months sales data analysed with NumPy.

```python
import numpy as np

products = ["Laptop", "Phone", "Tablet", "Watch", "Earbuds"]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = np.array([[120, 135, 150, 110, 160, 175],
                  [300, 280, 320, 350, 340, 390],
                  [ 90, 100,  85, 110, 120, 105],
                  [ 60,  75,  80,  70,  95, 100],
                  [200, 220, 210, 250, 230, 260]])

price = np.array([60000, 25000, 30000, 15000, 5000])
```

```python
# Task 1: shape of the sales table (products x months)
print(sales.shape)
```

```python
# Task 2: all Phone sales (row index 1)
print(sales[1])
```

```python
# Task 3: all March sales (column index 2)
print(sales[:, 2])
```

```python
# Task 4: total units sold per product
print(sales.sum(axis = 1))
```

```python
# Task 5: total units sold per month
print(sales.sum(axis = 0))
```

```python
# Task 6: best month (highest total units)
month_total = sales.sum(axis=0)
best = np.argmax(month_total)
print(months[best])
```

```python
# Task 7: best product (highest total units)
product_total = sales.sum(axis=1)
best_product = np.argmax(product_total)
print(products[best_product])
```

```python
# Task 8: entries above 200
print((sales > 200).sum())
```

```python
# Task 9: revenue table (units x price)
price_col = price.reshape(5, 1)
revenue = sales * price_col
print(revenue)
```

```python
# Task 10: revenue totals
print(revenue.sum(axis=1))
print(revenue.sum())
```
